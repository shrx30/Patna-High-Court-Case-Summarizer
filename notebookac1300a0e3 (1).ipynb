```python
import spacy
import requests
import json
import os

# Function to extract Named Entities using spaCy
def extract_named_entities(text):
    nlp = spacy.load("en_core_web_sm")  # Load the spaCy model
    doc = nlp(text)  # Process the text

    # Extract named entities (entities with their label)
    named_entities = [(ent.text, ent.label_) for ent in doc.ents]
    return named_entities

# Function to send named entities to Gemini API
def send_named_entities_to_gemini(named_entities, api_key):
    # Constructing input data for the Gemini API
    input_data = {
        "named_entities": named_entities  # List of named entities
    }

    # URL for the Gemini API (replace this with the actual URL)
    url = "https://api.gemini.example.com/proces"  # Update with actual Gemini API URL
    headers = {
        "Authorization": f"Bearer {api_key}",  # Bearer token authentication
        "Content-Type": "application/json"  # Ensure content type is JSON
    }

    try:
        # Send a POST request to the Gemini API
        response = requests.post(url, json=input_data, headers=headers)

        # Check if the request was successful
        if response.status_code == 200:
            # Parse the JSON response
            result = response.json()
            print("\nGemini Response:")
            print(json.dumps(result, indent=2))
            return result
        else:
            print(f"Error: {response.status_code} - {response.text}")
            return None
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
        return None

# Example usage
if __name__ == "__main__":
    # Sample input text
    text = "In the jurisdiction of Patna High Court, Laxman Ram filed a petition against Bihar State."

    # Extract named entities from the text
    named_entities = extract_named_entities(text)

    # Display extracted named entities
    print("Extracted Named Entities:")
    for entity in named_entities:
        print(f"{entity[0]} ({entity[1]})")

    # API key (use environment variable or secure storage for production use)
    api_key = os.getenv("AIzaSyCUNLHvwC__B7neqQ4jbJGyqJ1biyqgHoQ")  # Replace with your actual Gemini API key securely

    # Ensure the API key is set
    if not api_key:
        print("AIzaSyBRkXdsoUYfs_S6PHfRCT3GPPUxt6CjKSE")
    else:
        # Send named entities to Gemini API
        send_named_entities_to_gemini(named_entities, api_key)
```
